```python
# Colab Cell 3 — Basic stopword removal from a string (example_sent)
# -----------------------------------------------------------------
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

example_sent = """This is a sample sentence,
showing off the stop words filtration."""

# Build stopword set
stop_words = set(stopwords.words('english'))

# Tokenize
word_tokens = word_tokenize(example_sent)

# Option A: Remove stopwords (case-insensitive) but keep punctuation filtering optional
filtered_sentence = [w for w in word_tokens if w.lower() not in stop_words]

print("Original tokens:", word_tokens)
print("Filtered tokens (stopwords removed):", filtered_sentence)

# If you also want to remove punctuation:
filtered_no_punct = [w for w in filtered_sentence if w not in string.punctuation]
print("Filtered tokens (stopwords + punctuation removed):", filtered_no_punct)
```

```python
# Colab Cell 4 — Stopword removal on a file (text.txt -> filteredtext.txt)
# -----------------------------------------------------------------------
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

stop_words = set(stopwords.words('english'))

# Read the input file
with open('text.txt', 'r', encoding='utf-8') as f:
    text = f.read()

# Tokenize text (keeps punctuation tokens like ',' and '.')
tokens = word_tokenize(text)

# Filter tokens: remove stopwords (case-insensitive) and optionally punctuation
filtered_tokens = [t for t in tokens if t.lower() not in stop_words and t not in string.punctuation]

# Write output file (overwrite if exists)
with open('filteredtext.txt', 'w', encoding='utf-8') as out:
    # join with a single space for readability
    out.write(' '.join(filtered_tokens))

print("Wrote filteredtext.txt. First 200 chars preview:")
with open('filteredtext.txt', 'r', encoding='utf-8') as out:
    print(out.read()[:200])
```

```python
# Colab Cell 7 — Combined pipeline: normalize -> remove stopwords -> stem -> save
# -------------------------------------------------------------------------------
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
import string

stop_words = set(stopwords.words('english'))
ps = PorterStemmer()

def preprocess_text(text, remove_punctuation=True, lower=True, extra_stopwords=None):
    """
    Returns a list of processed tokens:
     - optional lowercasing
     - tokenization (word_tokenize)
     - stopword removal (case-insensitive)
     - optional punctuation removal
     - stemming using PorterStemmer
    extra_stopwords: iterable of additional words to treat as stopwords (strings)
    """
    if lower:
        text_proc = text.lower()
    else:
        text_proc = text

    tokens = word_tokenize(text_proc)

    # Merge standard stopwords with any extras
    sw = set(stop_words)
    if extra_stopwords:
        sw = sw.union({w.lower() for w in extra_stopwords})

    # Filter and stem
    processed = []
    for t in tokens:
        if remove_punctuation and t in string.punctuation:
            continue
        if t in sw:
            continue
        # stem the token
        stemmed = ps.stem(t)
        processed.append(stemmed)

    return processed

# demo: use the combined pipeline on the file text.txt and write to filtered_stemmed.txt
with open('text.txt', 'r', encoding='utf-8') as f:
    text_in = f.read()

# Add extra stopwords example: let's remove "sample" as well
processed_tokens = preprocess_text(text_in, remove_punctuation=True, lower=True, extra_stopwords=['sample'])

with open('filtered_stemmed.txt', 'w', encoding='utf-8') as fout:
    fout.write(' '.join(processed_tokens))

print("Wrote filtered_stemmed.txt. Preview:")
with open('filtered_stemmed.txt', 'r', encoding='utf-8') as fout:
    print(fout.read()[:400])
```
