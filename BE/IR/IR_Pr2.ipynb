```python
import re
import json
from collections import defaultdict
```

```python
# Example documents - you can replace, append, or load from files
documents = {
    "Document 1": "The quick brown fox jumped over the lazy dog.",
    "Document 2": "The lazy dog slept in the sun."
}

# Quick print to confirm
for doc_id, text in documents.items():
    print(doc_id, ":", text)
```

```python
def tokenize(text):
    """
    Lowercase + remove punctuation (keeps alphanumerics and apostrophes inside words),
    then split into tokens.
    """
    # Lowercase
    text = text.lower()
    # Replace non-word characters (except apostrophe) with space
    text = re.sub(r"[^\w']+", " ", text)
    # Split on whitespace and filter out empty tokens
    tokens = [t for t in text.split() if t]
    return tokens

# Demo tokenizer
print(tokenize("The quick, brown fox jumped over the lazy dog."))
print(tokenize("I don't want any unnecessary punctuation!"))
```

```python
def build_inverted_index(docs):
    """
    docs: dict mapping doc_id -> text
    returns: dict mapping term -> sorted list of doc_ids (no duplicates)
    """
    index = defaultdict(set)  # use set to avoid duplicate doc entries
    for doc_id, text in docs.items():
        tokens = tokenize(text)
        for token in tokens:
            index[token].add(doc_id)

    # convert sets to sorted lists for stable output
    index = {term: sorted(list(doc_ids)) for term, doc_ids in index.items()}
    return dict(sorted(index.items()))  # sort by term

# Build index
inverted_index = build_inverted_index(documents)

# Print index nicely
for term, doc_list in inverted_index.items():
    print(f"{term} -> {', '.join(doc_list)}")
```
